```python
# School Management System:
class Person:
  def __init__(self,name,ID):
    self.name = name
    self.ID = ID
class Teacher(Person):
  def info(self):
    print(f"Teacher's Name: {self.name}")
    print(f"Teeacher's ID: {self.ID}")
class Student(Person):
  def __init__(self,name,ID,Math,English,Urdu,Computer,Physics):
    super().__init__(name,ID)
    self.English = English
    self.Math = Math
    self.Urdu = Urdu
    self.Computer = Computer

    self.Physics = Physics
    self.total = English + Math + Urdu + Computer + Physics
    self.percentage = (self.total/500)*100

  def info(self):
    print(f"Student's Name: {self.name}")
    print(f"Student's ID: {self.ID}")
    print(f"Math Marks: {self.Math}")
    print(f"English Marks: {self.English}")
    print(f"Urdu Marks: {self.Urdu}")
    print(f"Computer Marks: {self.Computer}")
    print(f"Physics Marks: {self.Physics}")
    print(f"Total Marks:",self.total)
    print("Perecntage is :",self.percentage,"%")
class regularStudent(Student):
  def cal_grade(self):
    if self.percentage >=90:
      grade = "A"
    elif self.percentage >=80:
      grade = "B"
    elif self.percentage >=70:
      grade = "C"
    elif self.percentage >=60:
      grade = "D"
    else:
      grade = "F"
      print('You are Fail')
    print(f"Grade is: {grade}")
class scholarshipStudent(Student):
  def cal_grade(self):

    if self.percentage >=95:
      grade = "A"
    elif self.percentage >=85:
      grade = "B"
    elif self.percentage >=75:
      grade = "C"
    elif self.percentage >=65:
      grade = "D"
    else:
      grade = "F"
      print('You are Fail')
    print(f"Grade is: {grade}")

t = Teacher("Tahir",1234)
t.info()
print("-----------------------------------------")
students = [
    regularStudent("Ahmad",101,80,90,75,65,85),

    regularStudent("shehriyar",102,80,80,80,70,60),
    scholarshipStudent("Usman",103,77,67,87,97,40)
]

for student in students:
  student.info()
  student.cal_grade()
  print()



```
